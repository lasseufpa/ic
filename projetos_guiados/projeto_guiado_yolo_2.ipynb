# Practical Activity: Açaí Tree Annotation for YOLO Object Detection Using Roboflow

## 1. Objective of the Activity

The goal of this activity is to create a well-annotated image dataset for training a YOLO object detection model to detect and localize açaí palm trees in aerial or field images.

Students will use **Roboflow** to annotate images by drawing **bounding boxes** around visible açaí palm trees. The quality of the annotations is very important because YOLO learns directly from the labeled objects in the images.

Each student must annotate **50 images** by creating accurate bounding boxes around every visible açaí palm tree in the assigned images.

The main focus of the activity is to ensure:

- Accurate localization of açaí palm trees.
- Consistent use of the correct class label.
- Clean and reliable annotations suitable for YOLO training.
- High-quality dataset preparation for object detection.

---

## 2. Group Dynamics

This activity will be carried out individually (or in groups), depending on the instructor's arrangement.

Each student is responsible for annotating **50 images**.

Students must carefully inspect each image and create bounding boxes around all visible açaí palm trees.

---

## 3. Deliverables

Each student or group must submit:

- **Annotated Dataset:** The completed annotations in Roboflow.
- **Exported YOLO Dataset:** The dataset exported from Roboflow in YOLO format.
- **Short Technical Report:** A brief report describing the annotation process, challenges encountered, and how annotation quality was maintained.
- **Presentation:** A short technical explanation of the annotation work and dataset preparation process.

---

## 4. Annotation Requirements

The annotation must follow the guidelines below:

- **Object to Annotate:** Only açaí palm trees should be annotated.
- **Annotation Type:** Use rectangular bounding boxes.
- **Class Label:** Use a consistent class name: `acai`.
- **Bounding Box Accuracy:** The bounding box should tightly enclose the visible crown or body of the açaí palm tree without including too much background.
- **Multiple Objects:** If an image contains several açaí trees, each tree must be annotated separately.
- **Partially Visible Trees:** If an açaí tree is partially visible but still identifiable, it should be annotated.
- **Unclear Objects:** Do not annotate objects that are too unclear or cannot be confidently identified as açaí palm trees.
- **Consistency:** The same annotation rule must be applied across all 50 images.

---

## 5. Workflow Pipeline

The annotation workflow should follow the pipeline below:

1. **Image Access:** Create a Roboflow account and upload the assigned image dataset in Roboflow.
2. **Image Inspection:** Carefully examine each image to identify visible açaí palm trees.
3. **Bounding Box Creation:** Draw a bounding box around each açaí palm tree.
4. **Class Assignment:** Assign the correct class label to every bounding box.
5. **Quality Check:** Review each image to ensure no visible açaí tree is missed and no wrong object is labeled.
6. **Dataset Export:** Export the completed dataset in YOLO11 format.
7. **Model Preparation:** The annotated dataset will later be used to train a YOLO model for açaí tree detection and localization.

---

## 6. Report Requirements

The short report should include:

- **Introduction:** Briefly explain the purpose of the annotation activity.
- **Dataset Description:** State the number of images annotated and the type of images used.
- **Annotation Procedure:** Describe how Roboflow was used to create bounding boxes.
- **Annotation Rules:** Explain the criteria used to decide which trees were annotated.
- **Challenges:** Mention difficulties such as overlapping trees, unclear images, small objects, or partially visible trees.
- **Quality Control:** Explain how annotation accuracy and consistency were checked.
- **Conclusion:** Summarize the importance of good annotation for YOLO model training.

---

## 7. Instructions for Using Roboflow

Students should follow these steps:

1. Create or log in to a Roboflow account.
2. Create a project or workspace and add the instructor as a member.
3. Upload and access the assigned images.
4. Use the annotation tool to draw bounding boxes around açaí palm trees.
5. Assign the correct class label to each bounding box.
6. Save progress regularly.
7. Review all 50 images before final submission.
8. If permitted by the instructor, students may perform data augmentation by applying transformations such as rotation, horizontal or vertical flipping, brightness and contrast adjustments, scaling, and other techniques intended to improve dataset diversity and model robustness.
9. Export the dataset in YOLO11 format after completing the annotations.

---

## 8. Important Notes

- Poor annotations can reduce the accuracy of the YOLO model.
- Bounding boxes should be neat, tight, and consistent.
- Do not label non-açaí trees as açaí.
- Do not leave visible açaí palm trees unlabeled.
- The final dataset will be used for training and evaluating an object detection model.
